#while loop


```python
number=int(input("Enter a number to print its multiplication table using while loop: "))
counter=1
while counter<=10:
    print(f"{number} x {counter} = {number*counter}")
    counter+=1
    
```

```python
#to print even numbers till n
n=int(input("Enter a number to print even numbers till n: "))
i=0
while i<=n:
    print(i)
    i+=2 
```

```python
#to print even numbers till n using continue statement
n=int(input("Enter a number to print even numbers till n: "))
i=0
while i<=n:
    if i%2!=0:
        i+=1
        continue
    print(i)
    i+=1
     
```

```python
#define numbers till n( even odd prime) using for loop
n=int(input("Enter a number to define even, odd and prime numbers till n: "))   
for i in range(1, n+1):
    if i%2==0:
        print(f"{i} is even")
    else:
        print(f"{i} is odd")
    
    #check for prime number
    is_prime=True #is_prime is a flag variable used to indicate whether a number is prime or not
    for j in range(2, int(i**0.5)+1): 
        
        #why int(i**0.5)+1 because a larger factor of the number must be a multiple of a smaller factor that has been already checked
        #when we check for prime number we only need to check till the square root of that number
        #because if n=a*b and both a and b were greater than the square root of n, 
        #then a*b would be greater than n. So at least one of those factors must be less than or equal to the square root of n.
        #instead of doing int(i**0.5)+1 we can also check using integer is divisible by 2,3,5,7 etc but that will take more time but it looks like this
       
       if i%j==0:
            is_prime=False
            break
    if is_prime and i>1:
        print(f"{i} is prime")
            
    
```
